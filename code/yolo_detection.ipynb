## Object Detection Using YOLOv8

```python
from ultralytics import YOLO
import torch
import os
import cv2
import shutil
import pandas as pd

# Load YOLOv8 model
model = YOLO("yolov8n.pt")
```

```python
# Define input and output directories
input_dir = "../data/processed_data/train"
output_dir = "../data/yolo_processed_data"


# Processes all frames of individual videos in a their respective video folders using YOLO
def process_frames(video_folder):

    # Track the current video name
    video_name = os.path.basename(video_folder)

    # Create the name for the video output folder
    output_video_folder = os.path.join(output_dir, f"video_{video_name}")

    # Create the output folder directory for each video
    os.makedirs(output_video_folder, exist_ok = True)
    
    detection_data = []

    # Process each frame
    for frame_file in sorted(os.listdir(video_folder)):

        # Remove file extension (.jpg)
        frame_name = os.path.splitext(frame_file)[0]

        # Establish the path of the current frame
        frame_path = os.path.join(video_folder, frame_file)

        # Run the frame through the YOLO detection model
        results = model(frame_path)

        # Save the YOLO processed frame with bounding boxes
        output_frame_path = os.path.join(output_video_folder, f"frame_{frame_name}_yolo.jpg")

        # Save results in the current video folder if they exist
        if results:
            results[0].save(filename=output_frame_path)

        # Extract YOLO detections
        for result in results:
            for i, box in enumerate(result.boxes.xyxy.cpu().numpy()):

                # Coordinates of bounding box
                x_min, y_min, x_max, y_max = box[:4]

                # Object class index
                object_class = int(result.boxes.cls[i].item())  

                 # Confidence score
                confidence = float(result.boxes.conf[i].item()) 
                
                # Append all bounding box infomration to the detection_data list
                detection_data.append([frame_file, x_min, y_min, x_max, y_max, object_class, confidence])

    # Save detections (bounding boxes of objects) as a CSV in the current video folder
    df = pd.DataFrame(detection_data, columns = ["frame", "x_min", "y_min", "x_max", "y_max", "object_class", "confidence"])
    df.to_csv(os.path.join(output_video_folder, "detections.csv"), index = False)
```

```python
# Delete the output directory if it exists, then create a new one
if os.path.exists(output_dir):
    shutil.rmtree(output_dir)
os.makedirs(output_dir, exist_ok = True)

# Loop through all video folders
for video_folder in sorted(os.listdir(input_dir)):
    video_path = os.path.join(input_dir, video_folder)

    if os.path.isdir(video_path):
        process_frames(video_path)
```
